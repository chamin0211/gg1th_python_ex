반복문-for

> 단축키
- text 셸 -> code 셸 : [esc -> y] 
- code 셸 -> text 셸 : [esc -> m]
- 실행 : ctrl + enter
- 실행 + 다음 셸로 커서 이동: shift + enter
- 셸 삭제: esc -> dd
- 되돌리기: z
- 현재 셸의 위쪽에 셸 추가 : esc + a
- 현재 셸의 아래쪽에 셸 추가: esc + b

```python
nums = [273, 310, 4, 6, 200, 100, 20, 50, 40, 101]
len(nums)
```

```python
# 100이상의 숫자만 출력
for num in nums:
    # 100이상 조건 참인것만 출력
    if num >= 100:
        print(f"- 100이상의 수: {num}")
```

```python
# 100이상의 숫자, 100 미만인 숫자를 따로 새로운 리스트에 넣어서 각각 리스트로 출력
over_num = []
under_num = []
for num in nums:
    # 100이상 조건 참인것만 출력
    if num >= 100:
        # 100이상 리스트 추가
        over_num.append(num)
    else:
        # 100미만 리스트 추가
        under_num.append(num)

print(over_num)
print(under_num)
```

- for + dict

```python
# nums 의 숫자가 가각 몇번 반복하는지 dict로 만들어서 출력
nums = [1,2,6,7,4,3,2,1,5,9,4,9,7,2,1,3,4,5,6,7,5,6,7,5]
counter = {}
for num in nums:
    # 그 숫자가 제일 처음 나왔을 때, 키값 + 카운트
    if num not in counter:
        counter[num] = 1
    # 그 숫자가 제일 처음 안나왔을 때 카운트
    else:
        counter[num] += 1

print(counter)
    
```

```python
order = [2, 3, 5, 1]
box = 10
for orders in order:
    box -= orders
    if box <= 0:
        print("품절되었습니다.")
        break
    else:
        print(f"주문 처리 완료. 남은 재고는 {box}개 입니다.")
```
