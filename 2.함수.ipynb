## 함수 문제 1

```python
def mul(*values):
    sum = 1
    for i in values:
        sum *= i
    print(sum)
    print("-"*20)
    
```

```python
def comMul(num1, num2):
    # 1~100까지 공배수 구하기
    num = num1 * num2
    results = [] # 공배수를 추가하기
    result = 0
    n = 1
    while result <= 100:
        result = num * n
        if result < 100:
            results.append(result)
            n = n + 1
        else:
            break
    print(results)
    return result
```

```python
input_data = input("두 수 입력 예) 2, 3> ")
input_data
```

```python
num1, num2 = input_data.split(",")
num1, num2 = int(num1), int(num2)
```

```python
result = comMul(num1, num2)
```

```python
print(f"{num1}, {num2}의 공배수는 {result}")
```
